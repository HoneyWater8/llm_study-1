```python
import psycopg2 
```

```python
con = psycopg2.connect(
  database="dvdrental", user="postgres", password="p1234", host="localhost"
)
```

```python
sql = """
  select
    f1.title
    , count(f2.actor_id) as actor_cnt
  from film f1
  left join film_actor f2 
    on f1.film_id = f2.film_id
  group by f1.film_id
  ;
"""
```

```python
cur = con.cursor()
```

```python
cur.execute(sql)
```

```python
result = cur.fetchall()
```
